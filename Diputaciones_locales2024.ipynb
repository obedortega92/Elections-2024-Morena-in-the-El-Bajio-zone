```python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.style.use('ggplot')
print('Matplotlib version:', mpl.__version__)
```

```python
df_dipl = {
        'Estado': ['Aguascalientes', 'Guanajuato', 'Jalisco', 'Michoacán', 'Querétaro', 'San Luis Potosí', 'Zacatecas'],
        'Votos': [29.18, 33.26, 28.85, 29.54, 33.55, 19.77, 28.52]
}

df_dipl = pd.DataFrame(df_dipl)
print(df_dipl)
```

```python
df_dipl.set_index('Estado', inplace=True)
```

```python
print(df_dipl.dtypes)
```

```python
df_dipl = df_dipl.sort_values('Votos', ascending=True)

ax = df_dipl.plot(kind='barh', figsize=(9, 6), color='royalblue', legend=False)

plt.xlabel('Porcentaje de Votación')
plt.ylabel('Estado')

for index, value in enumerate(df_dipl['Votos']):
    label = f"{value:,.2f}"
    plt.text(value - (value * 0.05), index, label, va='center', color='white', ha='right')

plt.tight_layout()
plt.savefig('grafico_barras_apiladas.png', dpi=300, bbox_inches='tight')
plt.show()
```
