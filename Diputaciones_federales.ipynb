```python
import geopandas as gpd
import pandas as pd
import folium
import json
```

```python
shp_file = "/content/disfed2024.shp"
data = gpd.read_file(shp_file)
```

```python
if data.crs != "EPSG:4326":
    data = data.to_crs("EPSG:4326")
```

```python
df_dip = pd.read_csv('/content/Diputaciones 2024.csv')
df_dip['ID'] = df_dip['ID'].astype(str)
data['ID'] = data['ID'].astype(str)
```

```python
mexico_latitude = 23.6345
mexico_longitude = -102.5528
mexico_map = folium.Map(location=[mexico_latitude, mexico_longitude], zoom_start=4)
```

```python
data['unique_id'] = data['ENTIDAD'].astype(str) + "_" + data['ID'].astype(str)
df_dip['unique_id'] = df_dip['ENTIDAD'].astype(str) + "_" + df_dip['ID'].astype(str)
```

```python
geojson_data = json.loads(data.to_json())
```

```python
folium.Choropleth(
    geo_data=geojson_data,
    data=df_dip,
    columns=['unique_id', 'MV'],
    key_on='feature.properties.unique_id',
    fill_color='YlOrRd',
    fill_opacity=0.9,
    line_opacity=0.2,
    legend_name='Margen de Victoria de Morena en los distritos federales electorales',
    nan_fill_color='gray',
    reset=True
).add_to(mexico_map)
```

```python
mexico_map.save('mapa.html')
mexico_map
```
