```python
import numpy as np
import pandas as pd

import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.cm as cm

mpl.style.use('ggplot')

print('Matplotlib version: ', mpl.__version__)
```

```python
df_senado = pd.read_csv('/content/ELECCION_SENADO.csv')
df_senado.head()
```

```python
df_senado.set_index('ESTADOS', inplace=True)
df_senado.head()
```

```python
df_barra = df_senado.iloc[:,[-1]]
df_barra.head()
```

```python
df_barra = df_barra.sort_values(by=['PROMEDIO'], ascending=False)

df_barra.plot(kind='bar', figsize=(12, 8))
plt.xlabel('Estado')
plt.ylabel('Porcentaje de votos')
plt.show()
```

```python
df_barra = df_barra.sort_values(by=['PROMEDIO'], ascending=False)

ax = df_barra.plot(kind='bar', figsize=(12, 8))
plt.xlabel('Estado')
plt.ylabel('Porcentaje de votos')

for p in ax.patches:
    ax.text(p.get_x() + p.get_width()/2, p.get_height(), '{:.2f}'  .format(p.get_height()),
            fontsize=7, color='black', ha='center', va='bottom')

plt.show()
```

```python
target_states = ['AGUASCALIENTES', 'QUERETARO', 'JALISCO', 'GUANAJUATO', 'MICHOACAN', 'SAN LUIS POTOSI', 'ZACATECAS']
```

```python
ax = df_barra.plot(kind='bar', figsize=(12, 8), color='red', legend=False)

for i, bar in enumerate(ax.patches):
    if df_barra.index[i] in target_states:
        bar.set_color('blue')

plt.xlabel('Estado')
plt.ylabel('Porcentaje de votos')

for p in ax.patches:
    ax.text(p.get_x() + p.get_width()/2, p.get_height(), '{:.2f}'  .format(p.get_height()),
            fontsize=7, color='black', ha='center', va='bottom')

plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
