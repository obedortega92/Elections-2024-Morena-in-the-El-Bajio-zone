```python
import geopandas as gpd
import pandas as pd
import folium
import json
```

```python
df_mun = pd.read_csv('/content/Ayuntamientos 2024.csv')
```

```python
with open('/content/GeoJSON - Municipios.json') as file:
        mapa_mun = json.load(file)

municipios_encontrados = []
for feature in mapa_mun['features']:
    if 'properties' in feature and feature['properties'].get('cvegeoedo') == '01':
        clave_municipio = feature['properties'].get('cvegeomuni')
        nombre_municipio = feature['properties'].get('nom_mun')
        municipios_encontrados.append({'clave': clave_municipio, 'nombre': nombre_municipio})


if municipios_encontrados:
    print(f"Municipios encontrados en {'01'}:")
    for municipio in municipios_encontrados:
        print(f"Clave: {municipio['clave']}, Nombre: {municipio['nombre']}")
else:
    print(f"No se encontraron municipios para el estado '{'01'}'.")
```

```python
mexico_latitud = 23.6345
mexico_longitude = -102.5528
mexico_map = folium.Map(location=[mexico_latitud, mexico_longitude], zoom_start=4)
```

```python
folium.Choropleth(
    geo_data=mapa_mun,
    data=df_mun,
    columns=['cvegeomuni', 'votacion'],
    key_on='feature.properties.cvegeomuni',
    fill_color='YlOrRd',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Porcentaje de votación de Morena en las elecciones municipales',
    reset=True,
    nan_fill_color='gray'
).add_to(mexico_map)
```
